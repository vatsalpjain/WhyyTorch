```python
#Input image and kernel
img = np.array([
    [4, 5, 3, 10, 5, 12],
    [-4, -5, 6, 2, 1, 2],
    [7, 9, 9, 3, 2 ,16],
    [-1, -2, 3, 4, 5, -9],
    [1, 2, 3, -14, 5, -7],
    [1, 2, 3, 4, 5, 6]])

kernel =np.array([
    [1, 1], 
    [1, -1]])

bais = np.random.rand()
```

```python
bais
```

```python
# Feature map output
output = np.zeros((
    len(img) - len(kernel) + 1,
    len(img[0]) - len(kernel[0]) + 1
))

# Convolution operation
for i in range(len(img)-len(kernel)+1):
    for j in range(len(img[0])-len(kernel[0])+1):
        patch = img[i:i+len(kernel), j:j+len(kernel[0])]
        result = patch * kernel
        output[i][j] = np.sum(result) + bais

output = np.maximum(0, output)
```

```python
output
```

```python
# Max pooling operation
stride = 2
pool_size = 2
pooled_output = np.zeros((
    (output.shape[0] - pool_size) // stride + 1,
    (output.shape[1] - pool_size) // stride + 1
))

for i in range(0, output.shape[0] - pool_size + 1, stride):
    for j in range(0, output.shape[1] - pool_size + 1, stride):
        patch = output[i:i+pool_size, j:j+pool_size]
        pooled_output[i//stride][j//stride] = np.max(patch)
    
```

```python
# Multiple Filters 
num_filters = 5

kernels = np.random.randn(num_filters, 3, 3)
biases = np.random.randn(num_filters)
all_feature_maps = []
```

```python

for i in range(num_filters):
    kernel = kernels[i]
    bias = biases[i]
    
    output = np.zeros((
        len(img) - len(kernel) + 1,
        len(img[0]) - len(kernel[0]) + 1
    ))
    
    for j in range(len(img)-len(kernel)+1):
        for k in range(len(img[0])-len(kernel[0])+1):
            patch = img[j:j+len(kernel), k:k+len(kernel[0])]
            result = patch * kernel
            output[j][k] = np.sum(result) + bias
    
    output = np.maximum(0, output)
    
    pooled_output = np.zeros((
        (output.shape[0] - pool_size) // stride + 1,
        (output.shape[1] - pool_size) // stride + 1
    ))
    
    for j in range(0, output.shape[0] - pool_size + 1, stride):
        for k in range(0, output.shape[1] - pool_size + 1, stride):
            patch = output[j:j+pool_size, k:k+pool_size]
            pooled_output[j//stride][k//stride] = np.max(patch)
    
    print(f"Filter {i+1} Pooled Output:\n{pooled_output}\n")
    all_feature_maps.append(pooled_output)

all_feature_maps = np.stack(all_feature_maps, axis=-1)
```

```python
# A RGB image with 3 channels (Red, Green, Blue)
img = np.random.randn(6, 6, 3)

# Multiple Filters 
num_filters = 5

kernels = np.random.randn(num_filters, 3, 3, 3)  # 3 channels for RGB
biases = np.random.randn(num_filters)
all_feature_maps = []

for i in range(num_filters):
    kernel = kernels[i]
    bias = biases[i]
    
    output = np.zeros((
        img.shape[0] - kernel.shape[0] + 1,
        img.shape[1] - kernel.shape[1] + 1
    ))
    
    for j in range(img.shape[0]-kernel.shape[0]+1):
        for k in range(img.shape[1]-kernel.shape[1]+1):
            patch = img[j:j+kernel.shape[0], k:k+kernel.shape[1]]
            result = patch * kernel
            output[j][k] = np.sum(result) + bias
    
    output = np.maximum(0, output)
    
    pooled_output = np.zeros((
        (output.shape[0] - pool_size) // stride + 1,
        (output.shape[1] - pool_size) // stride + 1
    ))
    
    for j in range(0, output.shape[0] - pool_size + 1, stride):
        for k in range(0, output.shape[1] - pool_size + 1, stride):
            patch = output[j:j+pool_size, k:k+pool_size]
            pooled_output[j//stride][k//stride] = np.max(patch)
    
    print(f"Filter {i+1} Pooled Output:\n{pooled_output}\n")
    all_feature_maps.append(pooled_output)

all_feature_maps = np.stack(all_feature_maps, axis=-1)
```
